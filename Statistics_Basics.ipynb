IQ scores.

Data have an absolute zero.
Ratios between data points are meaningful.
Differences, addition, subtraction, multiplication, and division are valid operations.
Examples:
Height (e.g., 150 cm, 180 cm).
Weight (e.g., 50 kg, 75 kg).

4. What is a box plot, and what can it tell you about the distribution of data?
A box plot (also called a box-and-whisker plot) is a graphical representation of the distribution of a dataset. It visually summarizes key aspects of the data, including central tendency, spread, and potential outliers, using a simple rectangular "box" and "whiskers."

Components of a Box Plot
Box:

Represents the interquartile range (IQR), which is the range between the first quartile (Q1) and the third quartile (Q3).
The line inside the box indicates the median (second quartile, Q2).

What a Box Plot Tells You
Central Tendency:

The median (line inside the box) provides a measure of the dataset's central value.
Spread:

The IQR (Q3 - Q1) reflects the spread of the middle 50% of the data.
The length of the whiskers indicates overall variability.
Skewness:

If the median is closer to Q1 or Q3, the data may be skewed.
Right-skewed: Longer whisker or more outliers on the right side.
Left-skewed: Longer whisker or more outliers on the left side.
Outliers:

Points beyond the whiskers highlight potential outliers, indicating unusually high or low values.
Symmetry:

A symmetrical box plot indicates a balanced distribution; an asymmetrical one suggests skewness.
Advantages of a Box Plot
Concise summary: Displays multiple aspects of a dataset in a single visualization.
Outlier detection: Highlights extreme values effectively.
Comparison: Facilitates comparison between multiple datasets side by side.
